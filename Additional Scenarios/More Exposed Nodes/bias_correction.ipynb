```python
import logging
import ipywidgets as widgets
from IPython.display import display, clear_output, HTML

# Configure logging
logging.basicConfig(
    level=logging.INFO,
    format='%(asctime)s - %(levelname)s - %(message)s',
    handlers=[logging.StreamHandler()]
)
logger = logging.getLogger(__name__)

def add_node_with_edges(graph, node_name):
    """
    Thêm node với 3 chế độ:
    1. Manual: Nhập tay.
    2. Preset 1: Priority=0, Prob=0.32 (Topology 1).
    3. Preset 2: Priority=1, Prob=0.8 (Topology 2 - Web/Data Server).
    """
    graph.add_node(node_name)

    global node_dropdown, user_text, root_text, direction_dropdown, output, priority_text_widget, mode_dropdown
    output = widgets.Output()

    # ==========================================
    # LOGIC: MANUAL MODE (GIỮ NGUYÊN)
    # ==========================================
    def add_edge_manual(b):
        with output:
            current_priority = graph.nodes[node_name].get('priority', 'N/A')
            priority_msg = f"<b>Node:</b> '{node_name}' (Priority={current_priority}).<br><b>Adding Edges (Manual):</b>"
            clear_output(wait=True)
            display(HTML(priority_msg))

            try:
                target_node = node_dropdown.value
                user_prob = float(user_text.value)
                root_prob = float(root_text.value)
                direction = direction_dropdown.value

                if not (0.0 <= user_prob <= 1.0 and 0.0 <= root_prob <= 1.0):
                    display(HTML("<b>Error:</b> Probabilities must be in range [0.0, 1.0]."))
                    display_manual_widgets(priority_msg)
                    return

                if direction == "To Target":
                    graph.add_edge(node_name, target_node, user=user_prob, root=root_prob)
                    edge_msg = f"Edge: {node_name} -> {target_node}"
                elif direction == "From Target":
                    graph.add_edge(target_node, node_name, user=user_prob, root=root_prob)
                    edge_msg = f"Edge: {target_node} -> {node_name}"
                else:
                    graph.add_edge(node_name, target_node, user=user_prob, root=root_prob)
                    graph.add_edge(target_node, node_name, user=user_prob, root=root_prob)
                    edge_msg = f"Bidirectional: {node_name} <-> {target_node}"

                display(HTML(f"<b>Notification:</b> {edge_msg} added."))
                logger.info(f"{edge_msg} added manually.")
                display_manual_widgets(priority_msg)
            except ValueError:
                display(HTML("<b>Error:</b> Invalid input."))

    def stop_adding(b):
        with output:
            clear_output(wait=True)
            display(HTML(f"<b>Notification:</b> Finished adding edges for node '{node_name}'."))
            logger.info(f"Finished process for node '{node_name}'.")

    def display_manual_widgets(priority_msg):
        with output:
            clear_output(wait=True)
            display(HTML(priority_msg))
            existing_nodes = [n for n in graph.nodes() if n != node_name]
            if not existing_nodes:
                display(HTML("<b>Warning:</b> No other nodes to connect."))
                stop_btn = widgets.Button(description="Finish")
                stop_btn.on_click(stop_adding)
                display(stop_btn)
                return

            global node_dropdown, user_text, root_text, direction_dropdown
            node_dropdown = widgets.Dropdown(options=existing_nodes, description='Target:')
            user_text = widgets.FloatText(value=0.0, description='User Prob:')
            root_text = widgets.FloatText(value=0.0, description='Root Prob:')
            direction_dropdown = widgets.Dropdown(options=["To Target", "From Target", "Bidirectional"], description='Direction:')

            add_btn = widgets.Button(description="Add Edge")
            stop_btn = widgets.Button(description="Stop")
            add_btn.on_click(add_edge_manual)
            stop_btn.on_click(stop_adding)
            display(node_dropdown, user_text, root_text, direction_dropdown, add_btn, stop_btn)

    def set_priority_manual(b):
        with output:
            try:
                p = float(priority_text_widget.value)
                graph.nodes[node_name]['priority'] = p
                msg = f"<b>Notification:</b> Node '{node_name}' priority set to {p}."
                display_manual_widgets(f"{msg}<br><b>Now adding edges:</b>")
            except ValueError:
                display(HTML("<b>Error:</b> Invalid priority."))

    def display_priority_widget():
        with output:
            clear_output(wait=True)
            display(HTML(f"<b>Manual Mode:</b> Set priority for '{node_name}'"))
            global priority_text_widget
            priority_text_widget = widgets.FloatText(value=0.0, description='Priority:')
            btn = widgets.Button(description="Set & Add Edges")
            btn.on_click(set_priority_manual)
            display(priority_text_widget, btn)

    # ==========================================
    # LOGIC: PRESET 1 (Priority=0)
    # ==========================================
    def apply_preset_1(b):
        with output:
            clear_output(wait=True)
            graph.nodes[node_name]['priority'] = 0.0
            logger.info(f"Preset 1: Node '{node_name}' priority set to 0.0.")

            # Danh sách cứng của Preset 1
            preset_edges = [
                {"target": "Host 1", "dir": "out"}, {"target": "Pad", "dir": "out"},
                {"target": "Web Server", "dir": "out"}, {"target": "Host 2", "dir": "out"},
                {"target": "Host 3", "dir": "out"}, {"target": "Host 1", "dir": "in"},
                {"target": "Pad", "dir": "in"}
            ]

            prob = 0.32
            log_html = f"<b>Preset 1 Applied (Priority=0):</b><br><ul>"
            for edge in preset_edges:
                target = edge["target"]
                # Logic dir: out = Node -> Target; in = Target -> Node
                if edge["dir"] == "out":
                    graph.add_edge(node_name, target, user=prob, root=prob)
                    log_html += f"<li>{node_name} -> {target} (0.32)</li>"
                else:
                    graph.add_edge(target, node_name, user=prob, root=prob)
                    log_html += f"<li>{target} -> {node_name} (0.32)</li>"

            log_html += "</ul><b>Finished.</b>"
            display(HTML(log_html))

    # ==========================================
    # LOGIC: PRESET 2 (Priority=1) - UPDATED
    # ==========================================
    def apply_preset_2(b):
        with output:
            clear_output(wait=True)
            # 1. Set Priority = 1
            graph.nodes[node_name]['priority'] = 1.0
            logger.info(f"Preset 2: Node '{node_name}' priority set to 1.0.")

            # 2. Danh sách cứng của Preset 2 (Format y chang Preset 1)
            # Dữ liệu lấy từ hình ảnh: Web Server, Data Server, Pad, Host 1
            preset_edges = [
                {"target": "Web Server", "dir": "out"},
                {"target": "Data Server", "dir": "out"},
                {"target": "Pad", "dir": "in"},
                {"target": "Host 1", "dir": "in"},
                {"target": "Web Server", "dir": "in"}
            ]

            # Xác suất cố định là 0.8
            prob = 0.8
            log_html = f"<b>Preset 2 Applied (Priority=1):</b><br><ul>"

            for edge in preset_edges:
                target = edge["target"]
                if edge["dir"] == "out":
                    graph.add_edge(node_name, target, user=prob, root=prob)
                    log_html += f"<li>{node_name} -> {target} (0.8)</li>"
                else:
                    graph.add_edge(target, node_name, user=prob, root=prob)
                    log_html += f"<li>{target} -> {node_name} (0.8)</li>"

            log_html += "</ul><b>Finished.</b>"
            display(HTML(log_html))

    # ==========================================
    # START SCREEN
    # ==========================================
    def on_mode_confirm(b):
        mode = mode_dropdown.value
        if mode == 'Manual':
            display_priority_widget()
        elif mode == 'Preset 1':
            apply_preset_1(None)
        elif mode == 'Preset 2':
            apply_preset_2(None)

    def display_start_screen():
        with output:
            clear_output(wait=True)
            display(HTML(f"<h3>Adding Node: {node_name}</h3>Select Configuration:"))

            global mode_dropdown
            mode_dropdown = widgets.Dropdown(
                options=['Manual', 'Preset 1', 'Preset 2'],
                value='Manual',
                description='Mode:',
            )

            start_btn = widgets.Button(description="Start", button_style='primary')
            start_btn.on_click(on_mode_confirm)

            # Thông tin gợi ý
            info = widgets.HTML(
                value="<br><i><b>Preset 1:</b> P=0. Edges with 0.32 prob (Host1, Pad, Web...)"
                      "<br><b>Preset 2:</b> P=1. Edges with 0.8 prob (Web, Data, Pad...)</i>"
            )

            display(mode_dropdown, start_btn, info)

    display(output)
    display_start_screen()
```

```python
# Cell 6: Module Bias Correction cho Incremental Learning
import torch
import torch.nn as nn

class MultiHeadDQNWithBias(nn.Module):
    """
    Wrapper bao quanh MultiHeadDQN để xử lý sự chênh lệch Q-value
    giữa các node đã học (Old Nodes) và node mới thêm vào (New Nodes).
    """
    def __init__(self, base_model, old_num_nodes, new_num_nodes):
        super().__init__()
        self.model = base_model
        self.old_M = old_num_nodes
        self.new_M = new_num_nodes

        # Tham số học được để điều chỉnh phân phối Q-value của node mới
        # Alpha: Scaling factor (mặc định 1.0)
        # Beta: Bias shift (mặc định 0.0 - có thể set dương nhẹ để khuyến khích explore node mới)
        self.alpha = nn.Parameter(torch.tensor(1.0))
        self.beta = nn.Parameter(torch.tensor(0.0))

    def forward(self, x):
        # x: State Vector
        # q_raw: [Batch, N_honeypots, M_total_nodes]
        # Lưu ý: base_model phải có output layer kích thước M_total_nodes
        q_raw = self.model(x)

        # Tách Q-values của node cũ và node mới
        # Giả sử các node mới luôn được append vào cuối danh sách node
        q_old = q_raw[:, :, :self.old_M]
        q_new = q_raw[:, :, self.old_M:]

        # Áp dụng bias correction chỉ cho các node mới
        q_new_corrected = self.alpha * q_new + self.beta

        # Ghép lại để trả về format chuẩn [Batch, N, M]
        return torch.cat([q_old, q_new_corrected], dim=2)
```

```python
# Cell 8:
# Function to check if alpha is saturated
def is_alpha_saturated(alpha_history, k_consecutive, target_update_freq, alpha_threshold):
    """
    Check if the last k_consecutive episodes' alpha values are saturated.
    """
    num_alpha_updates = k_consecutive // target_update_freq + (1 if k_consecutive % target_update_freq else 0)
    if len(alpha_history) < num_alpha_updates + 1:
        return False
    recent_alphas = alpha_history[-(num_alpha_updates + 1):]
    differences = [(recent_alphas[i] - recent_alphas[i-1]) for i in range(1, len(recent_alphas))]
    return all(diff < alpha_threshold for diff in differences)
```

```python
# Cell 9: Khởi tạo Model & Optimizer cho Incremental Learning
import torch
import torch.optim as optim
from torch.optim.lr_scheduler import ReduceLROnPlateau

# --- 1. Hàm lấy mẫu hành động thăm dò (Đã sửa padding) ---
def sample_exploration_action_matrix(N, M_new, M_old, max_padding_nodes, exploration_counter, min_explorations=10):
    """
    Lấy mẫu hành động [N, MAX_NODES] trong đó ít nhất 1 node thuộc tập M_new.
    """
    new_action_tuples = []

    # Chỉ xét các node thực tế (0 đến M_new-1)
    # combinations giúp tránh trùng lặp (k1, k2) và (k2, k1) là như nhau
    all_tuples = itertools.combinations(range(M_new), N)

    for nodes_tuple in all_tuples:
        # Kiểm tra xem có ít nhất một node mới (>= M_old)
        is_new = any(node_idx >= M_old for node_idx in nodes_tuple)

        if is_new:
            sorted_tuple = tuple(sorted(nodes_tuple))
            new_action_tuples.append(sorted_tuple)

    if not new_action_tuples:
        # Trường hợp hiếm: Không có node mới nào? (M_new == M_old)
        # Fallback về random thường
        print("Cảnh báo: Không tìm thấy tổ hợp node mới nào. Fallback random.")
        return np.zeros((N, max_padding_nodes), dtype=np.float32), True

    # Chọn tuple dựa trên số lần đã thăm
    under_explored = [t for t in new_action_tuples if exploration_counter[t] < min_explorations]

    if under_explored:
        selected_tuple = random.choice(under_explored)
        exploration_done = False
    else:
        selected_tuple = random.choice(new_action_tuples)
        exploration_done = True # Đã khám phá đủ vùng này

    exploration_counter[selected_tuple] += 1

    # Tạo ma trận one-hot với kích thước PADDING chuẩn
    action_matrix = np.zeros((N, max_padding_nodes), dtype=np.float32)
    for i, node_idx in enumerate(selected_tuple):
        action_matrix[i, node_idx] = 1.0

    return action_matrix, exploration_done


# --- 2. Khởi tạo Model Wrapper (Bias Correction) ---
# Sử dụng dqn_model đã load ở Cell 7
# Lưu ý: M_new ở đây là số node thực tế (ví dụ 9), không phải padding
model = MultiHeadDQNWithBias(
    base_model=dqn_model,
    old_num_nodes=M_old,
    new_num_nodes=M_new
).to(device)

# --- 3. Cấu hình Optimizer ---
# Chiến lược:
# - Alpha, Beta: Học nhanh (LR lớn) để nhanh chóng cân bằng bias
# - Mạng DQN gốc: Học chậm (LR nhỏ) để không quên tri thức cũ (Catastrophic Forgetting)

# Nhóm tham số Bias
bias_params = [model.alpha, model.beta]
# Nhóm tham số Mạng DQN
network_params = list(model.model.parameters())

optimizer = optim.SGD([
    {'params': bias_params, 'lr': 0.18},   # Học bias nhanh
    {'params': network_params, 'lr': 0.001} # Finetune mạng chậm
], momentum=0.9, nesterov=True)

scheduler = ReduceLROnPlateau(optimizer, mode='max', factor=0.5, patience=30)

print(f"Model đã sẵn sàng.")
print(f" - Old Nodes: {M_old}")
print(f" - New Nodes Total: {M_new}")
print(f" - Max Padding: {MAX_NODES_CONFIG}")
print(f" - Optimizer: SGD (Bias LR=0.1, Network LR=0.001)")
```

```python
# Cell 10: Vòng lặp Huấn luyện với Bias Correction & Target Network
from copy import deepcopy

# --- 1. Cấu hình Hyperparameters ---
# Scheduler đã khai báo ở Cell 9 (nếu chưa thì khai báo lại ở đây)
# scheduler = ReduceLROnPlateau(optimizer, mode='min', factor=0.8, patience=5, verbose=True, min_lr=1e-5)
criterion = nn.SmoothL1Loss() # Huber Loss ổn định hơn MSE

replay_buffer = ReplayBuffer(capacity=10000)
target_update_freq = 10 # Cập nhật target net sau mỗi 10 episode
batch_size = 32 # Tăng batch size lên chút để ổn định gradient
num_episodes = 200
gamma = 0.90

# --- 2. Khởi tạo Target Network ---
# Copy toàn bộ cấu trúc (bao gồm cả Bias Wrapper)
target_model = deepcopy(model)
target_model.load_state_dict(model.state_dict())
target_model.eval() # Target net không cần tính gradient

# Logging vars
losses = []
alpha_history = []
beta_history = []
successes = 0
exploration_done = False
exploration_counter = defaultdict(int)
log_freq = 100
# Logic Bias Correction
k_consecutive = 30
alpha_threshold = 0.005
bias_correction_active = True
# Đóng băng toàn bộ trước
for param in model.parameters():
    param.requires_grad = False
# Unfreeze các phần cần học (FC1, Output Heads, Alpha, Beta)
model.model.fc1.weight.requires_grad = True
model.model.fc1.bias.requires_grad = True
for i in range(env.num_honeypots):
    model.model.output_heads[i].weight.requires_grad = True
    model.model.output_heads[i].bias.requires_grad = True
model.alpha.requires_grad = True
model.beta.requires_grad = True


print(f"\n--- Bắt đầu Huấn luyện (N={N_honeypots}, Nodes={M_new}, Pad={MAX_NODES_CONFIG}) ---")

for episode in range(1, num_episodes + 1):
    # Reset Env
    state = env.reset().to(device)
    done = False
    total_loss = 0
    episode_steps = 0

    while not done:
        # Flatten state: [M_new, D] -> [1, M_new*D]
        # Lưu ý: Input vào model phải luôn cố định size (đã padding trong Env)
        # Do Env trả về state đã padding sẵn rồi, ta chỉ cần flatten
        state_tensor_flat = state.flatten().unsqueeze(0).to(device)

        # --- A. Chọn Hành động ---
        if not exploration_done:
            # Lấy mẫu hành động tập trung vào node mới
            # Cần truyền MAX_NODES_CONFIG để output action matrix đúng kích thước padding
            action, exploration_done = sample_exploration_action_matrix(
                N_honeypots, M_new, M_old,
                max_padding_nodes=MAX_NODES_CONFIG,
                exploration_counter=exploration_counter,
                min_explorations=10
            )
        else:
            # Chọn theo Policy (Greedy)
            with torch.no_grad():
                q_values = model(state_tensor_flat) # [1, N, MAX_NODES]

                # Quan trọng: Masking các node padding (từ M_new trở đi)
                # Để agent không bao giờ chọn nhầm vào vùng hư vô
                if M_new < q_values.shape[2]:
                    q_values[:, :, M_new:] = -float('inf')

                # Hàm select_action cũ cần điều chỉnh để nhận Tensor 3D hoặc ta squeeze
                action = select_action_multi_head(q_values.squeeze(0), num_real_nodes=M_new)

        # --- B. Bước đi (Step) ---
        next_state, reward, done, path, captured = env.step(action)
        next_state = next_state.to(device)

        # Lưu vào buffer
        replay_buffer.push(state, action, reward, next_state, done)
        state = next_state
        episode_steps += 1

        if reward == 1:
            successes += 1

        # --- C. Huấn luyện (Training Step) ---
        if len(replay_buffer) >= batch_size:
            states, actions, rewards, next_states, dones = replay_buffer.sample(batch_size)

            states = states.to(device)
            actions = actions.to(device)
            rewards = rewards.to(device)
            next_states = next_states.to(device)
            dones = dones.to(device)

            # Flatten inputs
            states_flat = states.flatten(start_dim=1)
            next_states_flat = next_states.flatten(start_dim=1)

            # 1. Tính Q hiện tại
            # q_values_all: [Batch, N, MAX_NODES]
            q_values_all = model(states_flat)

            # Lấy giá trị Q của hành động đã chọn
            # actions là one-hot, nhân vào sẽ lọc ra giá trị Q cần tìm
            current_q_per_head = (q_values_all * actions).sum(dim=2) # [Batch, N]
            current_q_total = current_q_per_head.sum(dim=1) # [Batch] - Tổng Q của cả 3 honeypot

            # 2. Tính Target Q (Dùng Target Network)
            with torch.no_grad():
                next_q_values_all = target_model(next_states_flat)

                # Masking cho Target: Cực kỳ quan trọng để ổn định
                if M_new < next_q_values_all.shape[2]:
                    next_q_values_all[:, :, M_new:] = -float('inf')

                # Max Q cho mỗi đầu (Double DQN logic có thể áp dụng ở đây nếu muốn)
                next_q_values_per_head = next_q_values_all.max(dim=2)[0] # [Batch, N]
                next_q_total = next_q_values_per_head.sum(dim=1) # [Batch]

                # Bellman Equation
                targets = rewards.squeeze(1) + (1 - dones.squeeze(1)) * gamma * next_q_total

            # 3. Backprop
            loss = criterion(current_q_total, targets)

            optimizer.zero_grad()
            loss.backward()
            # Gradient Clipping để tránh bùng nổ
            torch.nn.utils.clip_grad_norm_(model.parameters(), 1.0)
            optimizer.step()

            total_loss += loss.item()

    # --- D. Cuối Episode ---
    avg_loss = total_loss / max(episode_steps, 1)
    if episode >= batch_size:
        losses.append(avg_loss)
        scheduler.step(avg_loss) # Bật lại nếu muốn LR decay

    # Cập nhật Target Network
    if episode % target_update_freq == 0:
        target_model.load_state_dict(model.state_dict())

        # Logging chi tiết
        current_lr = optimizer.param_groups[0]['lr']
        alpha_val = model.alpha.item()
        beta_val = model.beta.item()

        alpha_history.append(alpha_val)
        beta_history.append(beta_val)
        if episode % (log_freq) == 0:
            print(f"Ep {episode} | Loss: {avg_loss:.4f} | Alpha: {alpha_val:.3f} | Beta: {beta_val:.3f}  | DSP: {((successes/100)*100):.2f}%")
            successes = 0
        # --- Kiểm tra bão hòa Alpha (Dừng Bias Correction) ---
        # if bias_correction_active and is_alpha_saturated(losses, k_consecutive, target_update_freq, alpha_threshold):
        #     print(f"--> Alpha bão hòa ở Episode {episode}. Chuyển sang Fine-tuning toàn bộ mạng.")
        #
        #     # Tắt gradient của alpha, beta
        #     model.alpha.requires_grad = False
        #     model.beta.requires_grad = False
        #
        #     # Mở khóa toàn bộ mạng (nếu trước đó có đóng băng gì)
        #     # Ở chiến lược này ta đã mở ngay từ đầu, nhưng ta sẽ giảm LR để tinh chỉnh
        #     for param in model.model.parameters():
        #         param.requires_grad = True
        #
        #     # Re-init optimizer với LR nhỏ hơn cho toàn mạng
        #     optimizer = torch.optim.SGD(model.parameters(), lr=0.001, momentum=0.9)
        #     bias_correction_active = False

print(f"\n--- Hoàn tất Học Tiệm Tiến ---")
```
